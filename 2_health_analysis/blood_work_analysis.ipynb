```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm= ChatGoogleGenerativeAI(model= "gemini-3.5-flash")
```

```python
with open ("blood_work.txt", "r") as f:
    blood_report = f.read()

print(blood_report [:200])
```

```python
prompt = f"""
You're a medical assistance with 10 year or experience.

from the blood report below. Exact all valude and clasify as high and low or Normal based on the range next to it.

I need your reposne in this format 
- Test name | Status: HIGH, LOW, NORMAL | Refrence: range

Blood Report:
{blood_report}
"""

response = llm.invoke(prompt)
value = response.text

print(value)
```
