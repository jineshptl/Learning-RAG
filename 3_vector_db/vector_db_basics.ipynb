```python
import chromadb

# create a chroma db client
client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
```

```python
collection.add (
    ids = ["id1", "id2", "id3", "id4"],
    documents = [
        "Apple is leading in iPhone",
        "Tesla is leading in EV cars",
        "Apple are high in fiber",
        "space IPO is at record high",
    ]
)
```

```python
data = collection.get(
    include = ["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate (data["embeddings"]):
    print(f"documents {i+1}")
    print("text: ", data ["documents"] [i])
    print("Embedding Values: ", len(emb))
    print("First 10 values: ", emb[:20])
```

```python
result = collection.query (
    query_texts = ["this is a query related to elon musk"],
    n_results = 5
)

print(result)
```
