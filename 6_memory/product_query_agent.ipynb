```python
from dotenv import load_dotenv
#from Langchain_google_genai import ChatGoogleGenerativeAI

from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent

load_dotenv()
```

```python
PRODUCTS = {
    "wireless weadphone": {"price": 79.99, "description": "over the ear, wireless, 30 hours audion time"},
    "wmart watch": {"price": 199.99, "description" : "Tracks steps & heart rate"},
    "mechnical keyboard": {"price": 64.99, "description" : "Tenkyless, quite key press"},
    "laptop stand": {"price": 29.99, "description" : "adjustable, fits Macbook pro 14inch"},
 
}

@tool
def get_product(name: str) -> str:
    """Look up a product """
    p = PRODUCTS.get(name.lower())
    if not p:
        return f"Product not found. Available: {' ,'.join(PRODUCTS)}"
    return str(p)

llm = ChatGroq(model = "llama-3.3-70b-versatile", temperature = 0)

agent = create_agent(
    llm,
    tools = [get_product],
    system_prompt = "Act as an helpful product assistance for an online tech store"
)
```

```python
def ask (question : str):
    result = agent.invoke({"messages": [{"role": "user", "content": question}]})
    print (result["messages"][-1].content) 
```

```python
ask ("what is the price of Wireless Headphone")
```

```python
REVIEWS = {
    "wireless weadphone": {"reviews": 1400, "ratings": 3.8},
    "wmart watch": {"reviews": 800, "ratings": 2.0},
    "mechnical keyboard": {"reviews": 200, "ratings": 4.1},
    "laptop stand": {"reviews": 777, "ratings": 4.8},
}

@tool
def get_review(name: str) -> str:
    """Lookup product review by Product name. Return product name number of review and rating"""
    p = REVIEWS.get(name.lower())
    if not p:
        return f"Review not available for this product: {' ,'.join(REVIEWS)}"
    return str(p)
```

```python
llm = ChatGroq (model= "llama-3.3-70b-versatile", temperature=0)

agent2 = create_agent(
    llm,
    tools = [get_product, get_review],
    system_prompt = "Act as an helpful product assistance for an online tech store. "
)

def ask2 (question : str):
    result = agent.invoke({"messages": [{"role": "user", "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask2("What is the headphone price?")
```

```python
ask2("what was the name of the product in previous ask?")
```

```python
from langgraph.checkpoint.memory import InMemorySaver

llm = ChatGroq (model = "llama-3.3-70b-versatile", temperature =0)

agent3 = create_agent(
    llm,
    tools=[get_product, get_review],
    system_prompt = "Act as a helpful product assistance for an online tech store.",
    checkpointer = InMemorySaver(),
)

def ask3 (question :str):
    config = {"configurable":{"thread_id": "Session-1"}}
    result = agent3.invoke(
        {"messages":[{"role": "user", "content": question}]},
        config
    )
    
    print (result["messages"][-1].content)
```

```python
ask3("What is the headphone price?")
```

```python
ask3("what was the name of the product in previous ask?")
```
