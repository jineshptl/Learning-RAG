```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm= ChatGoogleGenerativeAI(model= "gemma-4-31b-it")

response = llm.invoke("how many moon does Jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "your helpful but make the resposne in 1 short line please in less than 14 words"],
    ["human", "how many moon does Jupiter have?"]
])
print(response.text)
```

Call LLM with GROQ Model

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()
```

```python
llm = ChatGroq(
    model = "llama-3.3-70b-versatile"
    # reasoning_format = "parsed"
)

messages = [
    (
        "system", 
        "your helpful but make the resposne in 1 short line please in less than 14 words",
    ),
    (
        "human",
        "how many moon does Jupiter have?"
    ),
]

ai = llm.invoke(messages)
print (ai.content)
```
